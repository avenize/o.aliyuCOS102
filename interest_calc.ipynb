```python
calc_type = float(input('Enter 1, 2 or 3 for simple interest, compund interest or annuity plan'))
print(f'{calc_type}')
while True:
    def calc_sys():
        
        match calc_type:
            case 1:
                print('Calculating Simple Interest')
                try:
                    p = float(input('Enter the principal'))
                except ValueError:
                    print('Invalid Number')
                try:
                    r = float(input('Enter the rate'))
                except ValueError:
                    print('Invalid Number')
                try:
                    t = float(input('Enter the time'))
                except ValueError:
                    print('Invalid Number')
                simple_interest = p * (1 + ((r/100) * t))
                rounded_simple_interest = round(simple_interest, 2)
                print(f'Simple Interest:{rounded_simple_interest}')
            
            case 2:
                print('Calculating Compound Interest')
                try:
                    p = float(input('Enter the principal'))
                except ValueError:
                    print('Inavlid Number')
                try:
                    r = float(input('Enter the rate'))
                except ValueError:
                    print('Invalid Number')
                try:
                    n = float(input('Enter the number'))
                except ValueError:
                    print('Invalid Number')
                try:
                    t = float(input('Enter the time'))
                except ValueError:
                    print('Invalid Number')
                compound_interest = p * ((1 + (r/n))**(n*t))
                rounded_compound_interest = round(compound_interest, 2)
                print(f'Compound Interest: {rounded_compound_interest}')

            case 3:
                print('Calculating annuity plan')
                try:
                    pmt = float(input('Enter the principal'))
                except ValueError:
                    print('Invalid Number')
                try:
                    rate = float(input('Enter the rate'))
                except ValueError:
                    print('Invalid Number')
                try:
                    n = float(input('Enter the number'))
                except ValueError:
                    print('Invalid Number')
                try:
                    t = float(input('Enter the time'))
                except ValueError:
                    print('Invalid Number')
                rate = rate/100
                annuity_plan = pmt * ((((1 + (rate/t))**(t * n)) - 1)/ (rate/t))
                rounded_annuity_plan = round(annuity_plan, 2)
                print(f'Annuity plan: {rounded_annuity_plan}')

            case _: 
                print('An error has occured')

    calc_sys()
    return_option = input('Do you want to rerun the program\n Type y or n').lower()
    if return_option != 'y':
        break      
```
